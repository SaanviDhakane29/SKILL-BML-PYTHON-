GATE LOGIC

```python
def AND(input1, input2):
    return input1 * input2

def OR(input1, input2):
    return 1 if input1 + input2 >= 1 else 0

def XOR(input1, input2):
    return 1 if input1 != input2 else 0

def main():
    print("Enter 1 for AND Gate")
    print("Enter 2 for OR Gate")
    print("Enter 3 for XOR Gate")

    choice = input("Your choice is : ")


    input1 = int(input("Enter first input (0 or 1): "))
    input2 = int(input("Enter second input (0 or 1): "))

    if choice == '1':
        print("Your choice is AND Gate")
        result = AND(input1, input2)
        print(f"Output: {result}")
    elif choice == '2':
        print("Your choice is OR Gate")
        result = OR(input1, input2)
        print(f"Output: {result}")
    elif choice == '3':
        print("Your choice is XOR Gate")
        result = XOR(input1, input2)
        print(f"Output: {result}")
    else:
        print("Invalid choice. Please enter 1, 2, or 3.")


main()
```

The above code is just logic. Now, we will code the training part that involves neural networks for each gate - AND, OR, XOR

```python
#training data for AND gate
x_train_data = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_train_data = np.array([0, 0, 0, 1])
import tensorflow as tf
import numpy as np

# Training data for XOR gate
# Inputs
X = np.array([[0,0], [0,1], [1,0], [1,1]], dtype=np.float32)

# Expected outputs
y = np.array([[0], [1], [1], [0]], dtype=np.float32)

# Define the model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_shape=(2,), activation='sigmoid'),  # Hidden layer with 2 neurons
    tf.keras.layers.Dense(1, activation='sigmoid')                     # Output layer
])

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['accuracy'])

# Train the model
model.fit(X, y, epochs=5000, verbose=0)  # verbose=0 to suppress output

# Evaluate and test
print("\nXOR Gate Results using Trained Neural Network:")
for inputs in X:
    prediction = model.predict(np.array([inputs]), verbose=0)
    output = 1 if prediction >= 0.5 else 0
    print(f"Input: {inputs} → Output: {output}")
```
